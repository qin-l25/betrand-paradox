```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import random
import math
```

```python
#将单位圆参数化（R和theta），对于R和theta分别均匀采样得到的结果(对应概率1/2)
sample_x = []
sample_y = []
for _ in range(3000):
    r = random.random()
    theta = random.random() * math.pi * 2.0
    x = r*math.cos(theta)
    y = r*math.sin(theta)
    sample_x.append(x)
    sample_y.append(y)
plt.scatter(sample_x, sample_y, s=3, color='blue', label='Data Points')
plt.axis('equal')
```

```python
sample_theta = []
for _ in range(3000):
    r = random.random()
    theta = random.random() * math.pi * 2.0
    x = r*math.cos(theta)
    y = r*math.sin(theta)
    alpha = math.acos(min(1.0, r))
    beta = math.atan2(y, x)
    theta1 = (beta + alpha) % (2 * math.pi)
    theta2 = (beta - alpha) % (2 * math.pi)
    sample_theta += [theta1, theta2]
sns.histplot(sample_theta, bins=30, color="dodgerblue", edgecolor="white")
plt.show()
```

```python
#在正方形当中进行拒绝采样(对应概率1/4)
sample_x = []
sample_y = []
for _ in range(3000):
    x = random.uniform(-1, 1)
    y = random.uniform(-1, 1)
    if x*x+y*y <= 1.00:
        sample_x.append(x)
        sample_y.append(y)
plt.scatter(sample_x, sample_y, s=3, color='red', label='Data Points')
plt.axis('equal')
```

```python
sample_theta = []
for _ in range(3000):
    x = random.uniform(-1, 1)
    y = random.uniform(-1, 1)
    if x*x+y*y <= 1.00:
        r = math.hypot(x, y)
        alpha = math.acos(min(1.0, r))
        beta = math.atan2(y, x)
        theta1 = (beta + alpha) % (2 * math.pi)
        theta2 = (beta - alpha) % (2 * math.pi)
        sample_theta += [theta1, theta2]
sns.kdeplot(sample_theta, color="dodgerblue", linewidth=2)
plt.show()                  
```

```python
#逆变换采样，同样得到“均匀”的结果
sample_x = []
sample_y = []
for _ in range(3000):
    eta1 = random.random()
    eta2 = random.random()
    r = math.sqrt(eta1)
    theta = 2 * eta2 * math.pi
    sample_x.append(r*math.cos(theta))
    sample_y.append(r*math.sin(theta))
plt.scatter(sample_x, sample_y, s=3, color='green', label='Data Points')
plt.axis('equal')
```

```python
sample_theta = []
for _ in range(3000):
    eta1 = random.random()
    eta2 = random.random()
    r = math.sqrt(eta1)
    theta = 2 * eta2 * math.pi
    x = r * math.cos(theta)
    y = r * math.sin(theta)
    alpha = math.acos(min(1.0, r))
    beta = math.atan2(y, x)
    theta1 = (beta + alpha) % (2 * math.pi)
    theta2 = (beta - alpha) % (2 * math.pi)
    sample_theta += [theta1, theta2]
sns.kdeplot(sample_theta, color="dodgerblue", linewidth=2)
plt.show() 
```

```python
#对应于最终概率 1/3 的采样方法
sample_x = []
sample_y = []
for _ in range(3000):
    theta = random.random() * 2 * math.pi
    phi = random.random() * 2 * math.pi
    x = math.sin(theta) + math.sin(phi)*math.cos(theta+phi)
    y = 1 - math.cos(theta) + math.sin(phi)*math.sin(theta+phi)
    sample_x.append(x)
    sample_y.append(y)
plt.scatter(sample_x, sample_y, s=3, color='orange', label='Data Points')
plt.axis('equal')
```

```python
sample_theta = []
for _ in range(3000):
    theta = random.random() * 2 * math.pi
    phi = random.random() * 2 * math.pi
    x = math.sin(theta) + math.sin(phi)*math.cos(theta+phi)
    y = 1 - math.cos(theta) + math.sin(phi)*math.sin(theta+phi)
    r = math.hypot(x, y - 1)
    alpha = math.acos(min(1.0, r))
    beta = math.atan2(y - 1, x)
    theta1 = (beta + alpha) % (2 * math.pi)
    theta2 = (beta - alpha) % (2 * math.pi)
    sample_theta += [theta1, theta2]
sns.kdeplot(sample_theta, color="dodgerblue", linewidth=2)
plt.show() 
```
